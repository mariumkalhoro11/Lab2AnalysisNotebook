# Lab 2: Mayo Clinic PBC Trial Analysis

## Research Question
How do baseline bilirubin levels compare between the two treatment groups in the Mayo Clinic PBC trial?

## Objective
The goal of this project is to analyze a public health dataset using Python. I will load the data, clean and organize it, calculate summary statistics, and create a graph comparing bilirubin levels between treatment groups.

```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python

# Load the Mayo Clinic PBC dataset directly from its online source
url = "https://vincentarelbundock.github.io/Rdatasets/csv/survival/pbc.csv"

df = pd.read_csv(url)

# Show the first five rows
print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

df.head()
```

```python

# Explore the dataset
print("Column names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

print("\nSummary statistics:")
display(df.describe())
```

```python

# Step 8: Analyze bilirubin levels by treatment group

# Keep patients who were part of the randomized trial
trial = df[df["trt"].isin([1, 2])].copy()

# Remove missing bilirubin values
trial = trial.dropna(subset=["bili"])

# Give the treatment groups readable names
trial["treatment"] = trial["trt"].map({
    1: "D-penicillamine",
    2: "Placebo"
})

# Calculate summary statistics for each group
summary = (
    trial.groupby("treatment")["bili"]
    .agg(
        count="count",
        mean="mean",
        median="median",
        minimum="min",
        maximum="max"
    )
    .reset_index()
)

print("Bilirubin Levels by Treatment Group:")
display(summary)
```
